## Паттерн: Одиночка (Singleton)

new — это специальный метод, который вызывается перед init.
Мы сохраняем единственный экземпляр (cls._instance).

```python
class Singleton:
    _instance = None

    def __new__(cls):
        if cls._instance is None:
            print("Создание нового экземпляра Singleton")
            cls._instance = super(Singleton, cls).__new__(cls)
        return cls._instance

    def some_business_logic(self):
        print("Выполнение бизнес-логики Singleton")

# Пример использования:
singleton1 = Singleton()
singleton2 = Singleton()

singleton1.some_business_logic()
print(f"singleton1 is singleton2: {singleton1 is singleton2}")  # True
```

## Паттерн: Команда (Command)

Команды (объекты) инкапсулируют действия.
RemoteControl может управлять любыми командами, не зная, что именно делает команда.

```python
from abc import ABC, abstractmethod

# Интерфейс команды
class Command(ABC):
    @abstractmethod
    def execute(self):
        pass

# Получатель (Receiver)
class Light:
    def turn_on(self):
        print("Свет включён")

    def turn_off(self):
        print("Свет выключен")

# Конкретные команды
class TurnOnCommand(Command):
    def __init__(self, light):
        self.light = light

    def execute(self):
        self.light.turn_on()

class TurnOffCommand(Command):
    def __init__(self, light):
        self.light = light

    def execute(self):
        self.light.turn_off()

# Инициатор (Invoker)
class RemoteControl:
    def submit(self, command):
        command.execute()

# Пример использования:
light = Light()
on_command = TurnOnCommand(light)
off_command = TurnOffCommand(light)

remote = RemoteControl()
remote.submit(on_command)
remote.submit(off_command)
```

## Паттерн: Адаптер (Adapter)

Adapter наследует Target и содержит Adaptee.
Он преобразует интерфейс Adaptee к ожидаемому формату

```python
# Целевой интерфейс
class Target:
    def request(self):
        return "Target: стандартный интерфейс запроса"

# Класс с несовместимым интерфейсом
class Adaptee:
    def specific_request(self):
        return ".тсеревир тофяпон ананйЗ"

# Адаптер
class Adapter(Target):
    def __init__(self, adaptee):
        self.adaptee = adaptee

    def request(self):
        result = self.adaptee.specific_request()
        return f"Adapter: (перевод) {result[::-1]}"

# Пример использования:
def client_code(target):
    print(target.request())

# Клиент работает с Target
print("Клиент работает с Target:")
target = Target()
client_code(target)

# Но теперь он также может работать с Adaptee через Adapter
print("\nКлиент работает с Adaptee через Adapter:")
adaptee = Adaptee()
adapter = Adapter(adaptee)
client_code(adapter)
```

## Паттерн: Фасад (Facade)

Клиенту не нужно знать детали CPU, Memory и HardDrive.
ComputerFacade инкапсулирует весь сложный процесс запуска.

```python
# Подсистема 1
class CPU:
    def freeze(self):
        print("CPU: заморозка")

    def jump(self, position):
        print(f"CPU: переход к {position}")

    def execute(self):
        print("CPU: выполнение")

# Подсистема 2
class Memory:
    def load(self, position, data):
        print(f"Память: загрузка {data} в {position}")

# Подсистема 3
class HardDrive:
    def read(self, lba, size):
        return f"данные с позиции {lba}, размер {size}"

# Фасад
class ComputerFacade:
    def __init__(self):
        self.cpu = CPU()
        self.memory = Memory()
        self.hard_drive = HardDrive()

    def start(self):
        print("Запуск компьютера через фасад:")
        self.cpu.freeze()
        data = self.hard_drive.read(100, 64)
        self.memory.load(0, data)
        self.cpu.jump(0)
        self.cpu.execute()

# Пример использования:
computer = ComputerFacade()
computer.start()
```

## Паттерн: Состояние (State)

Поведение объекта (Context) зависит от текущего состояния (State).
Объект состояния сам определяет, в какое состояние перейти дальше.

```python
from abc import ABC, abstractmethod

# Интерфейс состояния
class State(ABC):
    @abstractmethod
    def handle(self, context):
        pass

# Контекст
class Context:
    def __init__(self, state):
        self._state = state

    def set_state(self, state):
        print(f"Контекст: переход к {state.__class__.__name__}")
        self._state = state

    def request(self):
        self._state.handle(self)

# Конкретные состояния
class ConcreteStateA(State):
    def handle(self, context):
        print("Состояние A обрабатывает запрос.")
        context.set_state(ConcreteStateB())

class ConcreteStateB(State):
    def handle(self, context):
        print("Состояние B обрабатывает запрос.")
        context.set_state(ConcreteStateA())

# Пример использования:
context = Context(ConcreteStateA())

# Последовательные вызовы
context.request()
context.request()
context.request()
```
